```python
import requests
from bs4 import BeautifulSoup

blog_url = "https://www.python.org/downloads/release/python-3130/?utm_source=chatgpt.com"

response = requests.get(blog_url)

print("Status code:", response.status_code)
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text, "html.parser")

# Remove unnecessary elements
for element in soup(["script", "style", "nav", "footer"]):
    element.decompose()

# Extract text
article_text = soup.get_text(separator=" ", strip=True)

print(article_text[:3000])
```

```python
prompt = f"""
You are a professional podcast script writer.

Convert the following blog article into an engaging 2-person podcast conversation.

Requirements:
- Create two speakers: Host and Guest
- Make it natural and conversational
- Explain technical terms simply
- Cover the important points from the article
- Do not invent facts that are not in the article
- Keep the script around 3 minutes

BLOG ARTICLE:
{article_text}
"""

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=prompt
)

podcast_script = response.text

print(podcast_script)
```
